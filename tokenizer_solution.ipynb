```python
from transformers import AutoTokenizer
```

```python

text = "I can easily tokenize text with Hugging Face"

tokenizer = AutoTokenizer.from_pretrained("cardiffnlp/twitter-roberta-base-sentiment")

encoded_input = tokenizer(text)

input_ids = encoded_input["input_ids"]

print(input_ids)
```

```python
tokenizer.decode(0)
```

```python
tokenizer.decode(100)
```

```python
tokenizer.decode(2)
```

```python

first_token = tokenizer.decode([input_ids[0]])
last_token = tokenizer.decode([input_ids[-1]])

print("First token:", first_token)
print("Last token:", last_token)
```

```python
encoded_input = tokenizer(
    text,
    max_length=20,
    padding="max_length",
    truncation=True
)

print("Input IDs:", encoded_input["input_ids"])
print("Attention Mask:", encoded_input["attention_mask"])
```

```python
print("Padding token:", tokenizer.pad_token)
print("Padding token ID:", tokenizer.pad_token_id)
```
