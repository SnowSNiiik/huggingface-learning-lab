```python
from transformers import pipeline
```

```python
sentiment_pipeline = pipeline("sentiment-analysis")
```

```python
comment_1 = "My experience so far has been fantastic!"
comment_2 = "The product is okay I guess"
comment_3 = "Your support team is useless"
```

```python
sentiment_pipeline(comment_1)
sentiment_pipeline(comment_2)
sentiment_pipeline(comment_3)
```

```python
#This is great!
comment_1_hindi = "यह भी खूब रही!"

sentiment_pipeline(comment_1_hindi)
```

```python
sentiment_pipeline_specified = pipeline("sentiment-analysis", model="tabularisai/multilingual-sentiment-analysis")
```

```python
sentiment_pipeline_specified(comment_1_hindi)
```
